```python
#W6A4
n=input().split()
d={}
for pair in n:
    key,value = pair.split(":")
    if key not in d:
       d[key]=[value]
    else:
       d[key].append(value)

print(d)
```

```python
#W6A5
n=map(int,input("nhap cac so nguyen: ").split())
d={
   'positives' : 0,
   'negative' : 0,
   'zero' : 0
}

for x in n:
    if x > 0:
       d['positives'] +=  1
    elif x < 0:
       d['negative'] += 1
    else:
       d['zero'] += 1

print(d)
```

```python
#W6A10
n = list(map(int,input().split()))
k = int(input())
a=[]
for x in n:
    y = k - x
    if y in n and x < y:
        a.append((x, y))

print(a)
```

```python
#W6A12
n = list(map(int,input("nhap cac so nguyen: ").split()))
d = {}
for x in n:
    key = x
    if key in d:
        d[key] += 1 
    else:
        d[key] = 1 

b=list(d.items())
b.sort(key = lambda x : (-x[1], x[0]))
print(b[0][0])
```

```python
#W6A14
ds1 = list(map(int,input().split()))
ds2 = list(map(int, input().split()))
ds=[]
for i in range(len(ds1)):
    if ds1[i] in ds2:
        ds.append(ds1[i])
ds=list(set(ds))
ds.sort()
print(ds)
```
