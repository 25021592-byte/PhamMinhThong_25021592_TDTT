```python
#W7A1
def binary_search(a,l,r,x):
    while l <= r:
        m = (l + r) // 2
        if a[m] == x:
            return m
        elif a[m] < x:
            l = m + 1
        else:
            r = m - 1
    return -1

a = list(map(int,input().split()))
k = int(input())
print(binary_search(a,0,len(a)-1,k))
```

```python
#W7A2
def count_occurrences(arr, x) -> int:
    dem = 0
    for i in arr:
        if i == x:
            dem += 1
    return dem 

arr = list(map(int,input().split()))
x = int(input())
print(count_occurrences(arr,x))
```

```python
#W7A3
def selection_sort(arr):
    n = len(arr)
    for i in range(n):
        min_idx = i
        for j in range(i + 1,n):
            if arr[j] < arr[min_idx]:
                min_idx = j
        
        arr[i], arr[min_idx] = arr[min_idx], arr[i]
    return arr

def bubble_sort(arr):
    n = len(arr)
    for i in range(n):
        swapped = False
        for j in range(0,n - i - 1):
            if arr[j] > arr[j + 1]:
                arr[j], arr[j +1 ] = arr[j + 1], arr[j]
                swapped = True
        if not swapped:
            break
    return arr

arr = list(map(int,input().split()))
print(*selection_sort(arr))
print(*bubble_sort(arr))
```

```python
#W7A4
n = list(map(int,input().split()))
d = {}
pos = {}
for i in range(len(n)):
    if n[i] in d:
       d[n[i]] += 1
    else:
        d[n[i]] = 1
        pos[n[i]] = i
b = list(d.items())
b.sort(key = lambda x : (-x[1], pos[x[0]]))
print(b[0][0], b[0][1])
```

```python
#W7A7
n = list(map(int,input().split()))
dp = [1]*len(n)

for i in range(len(n)):
    for j in range(i):
        if n[j] < n[i]:
            dp[i] = max(dp[i], dp[j] + 1)

print(max(dp)) 
```

```python
#W7A11
def tansuattu(cau,tu):
    d = {}
    for ch in cau:
        d[ch] = d.get(ch,0) + 1
    return d.get(tu, 0)

cau = input().split()
tu = input()
print(tansuattu(cau,tu))
```

```python
#W7A12
def sap_xep_nhanh():
    labels = ['A', 'B', 'C', 'D']
    phuong_an = []
    print("Nhập 4 phương án (mỗi phương án 1 dòng):")
    for i in range(4):
        try:
            text = input().strip()
            phuong_an.append((text, labels[i]))
        except EOFError:
            break
    
    phuong_an.sort(key=lambda x: x[0])
    ket_qua = [item[1] for item in phuong_an]

    print("Thứ tự đúng:")
    print(" ".join(ket_qua))

if __name__ == "__main__":
    sap_xep_nhanh()
```

```python
#W7A13
def longest_harmonious_subsequence(arr):
    freq = {}
    for x in arr:
        freq[x] = freq.get(x, 0) + 1
    max_len = 0
    for x in freq:
        if x + 1 in freq:
            max_len = max(max_len, freq[x] + freq[x + 1])
    return max_len

arr = list(map(int, input().split()))
print(longest_harmonious_subsequence(arr))
```

```python
#W7A14
def timso7(arr):
    for x in range(len(arr)):
        if arr[x] == 7:
            return x
    return "Not found"

n = int(input())
arr = []
for i in range(0,n):
    temp = int(input())
    arr.append(temp)
    
print(timso7(arr))
```
